## Clustering

Clustering is where we group similar data points together without prior labels. This process helps in identifying patterns or structures within the data.

### K-means clustering

* One of the simplest clustering methods
* The algorithm tries to identify the centre of each cluster

```python
# K-means with Scikit-Learn
import sklearn.cluster as skl_cluster
```

We'll use use Scikit-Learn's built-in "random data blob generator" for this exercise.

```python
import sklearn.datasets as skl_datasets
```

Create blobs using the make_blobs function

```python
import matplotlib.pyplot as plt

#Lets define some functions here to avoid repetitive code
def plots_labels(data, labels):
    tx = data[:, 0]
    ty = data[:, 1]

    fig = plt.figure(1, figsize=(4, 4))
    plt.scatter(tx, ty, edgecolor='k', c=labels)
    plt.show()

def plot_clusters(data, clusters, Kmean):
    tx = data[:, 0]
    ty = data[:, 1]
    fig = plt.figure(1, figsize=(4, 4))
    plt.scatter(tx, ty, s=5, linewidth=0, c=clusters)
    for cluster_x, cluster_y in Kmean.cluster_centers_:
        plt.scatter(cluster_x, cluster_y, s=100, c='r', marker='x')
    plt.show()
```

Create the clusters

```python
data, cluster_id = skl_datasets.make_blobs(n_samples=400, cluster_std=0.75, centers=4, random_state=1)
plots_labels(data, cluster_id)
```

Fit the K-Means Model

```python
Kmean = skl_cluster.KMeans(n_clusters=4)
Kmean.fit(data)
clusters = Kmean.predict(data)
```

```python
plot_clusters(data, clusters, Kmean)
```

Pulling all the code into a single code block:

```python
import sklearn.cluster as skl_cluster
import sklearn.datasets as skl_datasets
import matplotlib.pyplot as plt

data, cluster_id = skl_datasets.make_blobs(n_samples=400, cluster_std=0.75, centers=4, random_state=1)

Kmean = skl_cluster.KMeans(n_clusters=4)
Kmean.fit(data)
clusters = Kmean.predict(data)

plot_clusters(data, clusters, Kmean)
```

Overlapping Clusters

```python
data, cluster_id = skl_datasets.make_blobs(n_samples=4000, cluster_std=3, centers=4, random_state=1)

Kmean = skl_cluster.KMeans(n_clusters=4)
Kmean.fit(data)
clusters = Kmean.predict(data)

plot_clusters(data, clusters, Kmean)
```

How many clusters should we look for?

We can modify our code to loop through values of k to see which (if any) of the results look reasonable.

```python
for cluster_count in range(2,11):
    Kmean = skl_cluster.KMeans(n_clusters=cluster_count)
    Kmean.fit(data)
    clusters = Kmean.predict(data)
    plt.scatter(data[:, 0], data[:, 1], s=5, linewidth=0,c=clusters)
    for cluster_x, cluster_y in Kmean.cluster_centers_:
        plt.scatter(cluster_x, cluster_y, s=100, c='r', marker='x')
        # give the graph a title with the number of clusters
        plt.title(str(cluster_count)+" Clusters")
    plt.show()
```

Elbow method

Plot the "inertia" (sum of squared distances) for *K* = 1 to 10. The elbow in the graph is usually a good *K* to use.

```python
import matplotlib.pyplot as plt
import sklearn.cluster as skl_cluster
import sklearn.datasets as skl_datasets

# Generate data
data, _ = skl_datasets.make_blobs(n_samples=500, centers=4, cluster_std=0.6, random_state=0)

inertias = []
K_values = range(1, 11)

for k in K_values:
    kmean = skl_cluster.KMeans(n_clusters=k)
    kmean.fit(data)
    inertias.append(kmean.inertia_)  # Sum of squared distances

# Plot the elbow graph
plt.plot(K_values, inertias, marker='o')
plt.xlabel('Number of clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal K')
plt.show()
```

### Spectral Clustering

Spectral clustering uses relationships among observations to transform the clustering problem into a representation where those groups can be separated more effectively.

```python
#Create concentric circles

import sklearn.datasets as skl_data

circles, circles_clusters = skl_data.make_circles(n_samples=400, noise=.01, random_state=0)
plots_labels(circles, circles_clusters)
```

```python
model = skl_cluster.SpectralClustering(n_clusters=2, affinity='nearest_neighbors', assign_labels='kmeans')
```

```python
labels = model.fit_predict(circles)
plots_labels(circles, labels)
```

Combining code and comparing with k means clustering

```python
import sklearn.cluster as skl_cluster
import sklearn.datasets as skl_data

circles, circles_clusters = skl_data.make_circles(n_samples=400, noise=.01, random_state=0)

# cluster with kmeans
Kmean = skl_cluster.KMeans(n_clusters=2)
Kmean.fit(circles)
clusters = Kmean.predict(circles)

# plot the data, colouring it by cluster
plot_clusters(circles, clusters, Kmean)

# cluster with spectral clustering
model = skl_cluster.SpectralClustering(n_clusters=2, affinity='nearest_neighbors', assign_labels='kmeans')
labels = model.fit_predict(circles)
plots_labels(circles, labels)
```

### DBSCAN

DBSCAN groups together points that are close to each other and labels isolated points as noise. Unlike K-Means, you don't have to choose K!

```python
from sklearn.cluster import DBSCAN
import sklearn.datasets as skl_data
import matplotlib.pyplot as plt

# Generate concentric circles
circles, _ = skl_data.make_circles(n_samples=400, noise=0.02, factor=0.5, random_state=1)

# DBSCAN clustering
model = DBSCAN(eps=0.1, min_samples=5)
labels = model.fit_predict(circles)

# Plot the clusters
plt.scatter(circles[:, 0], circles[:, 1], c=labels, cmap='viridis', s=20)
plt.title('DBSCAN Clustering')
plt.show()
```

# Part 4: Neural Networks

Basic Structure:
* Input Layer: Where the data goes in
* Hidden Layers: Where the calculations + pattern-finding happens
* Output Layer: Where the result comes out

Import libraries and load data

```python
import numpy as np
import tensorflow as tf # Framework for building neural networks
from tensorflow import keras # Simplified tools inside TensorFlow
import matplotlib.pyplot as plt

# load data
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize
x_train = x_train / 255.0
x_test = x_test / 255.0

# Show the first image in the training set
plt.imshow(x_train[0], cmap='gray')
plt.title(f'Label: {y_train[0]}')
plt.show()
```

Build a neural network

```python
# Build model
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),  # Flatten image
    keras.layers.Dense(128, activation='relu'),  # Hidden layer
    keras.layers.Dense(10, activation='softmax') # Output layer (10 digits)
])

# Compile model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(x_train, y_train, epochs=5, validation_split=0.2)
```

Evaluate and make predictions

```python
# Evaluate on test data
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f'Test Accuracy: {test_acc:.2f}')
```

```python
# Predict on test set
predictions = model.predict(x_test)

# Get predicted label for first test image
predicted_label = np.argmax(predictions[0])
actual_label = y_test[0]

# Display prediction vs. actual
plt.imshow(x_test[0], cmap='gray')
plt.title(f'Predicted: {predicted_label}, Actual: {actual_label}')
plt.show()
```
